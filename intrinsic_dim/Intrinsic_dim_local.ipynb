```python
import os

import numpy as np
import pandas as pd
from tqdm import tqdm

from intrinsic_dim import compute_intrinsic_dim_local
```

```python
EMBEDDINGS_PATH = '/home/jovyan/shares/SR003.nfs2/recsys_data/recsys_challenge_2025/stage2/unsupervised_gru_base/'
```

```python
embedding_dirs = [f for f in os.listdir(EMBEDDINGS_PATH) if os.path.isdir(os.path.join(EMBEDDINGS_PATH, f))]
len(embedding_dirs)
```

```python

results_local = {}
estimator_names = ['MLE', 'MOM']

for embedding_dir in tqdm(embedding_dirs):
    emb = np.load(os.path.join(EMBEDDINGS_PATH, embedding_dir, 'embeddings.npy'))
    results_local[embedding_dir] = compute_intrinsic_dim_local(
        emb, estimator_names=estimator_names, n_neighbors=100, n_jobs=-1)

results_local = pd.DataFrame(results_local).T
```
