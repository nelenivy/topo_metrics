```python
import os

import numpy as np
import pandas as pd
from tqdm import tqdm

from intrinsic_dim import compute_intrinsic_dim_global
```

```python
EMBEDDINGS_PATH = '/home/jovyan/shares/SR003.nfs2/recsys_data/recsys_challenge_2025/stage2/unsupervised_gru_base/'
```

```python
embedding_dirs = [f for f in os.listdir(EMBEDDINGS_PATH) if os.path.isdir(os.path.join(EMBEDDINGS_PATH, f))]
len(embedding_dirs)
```

```python
emb = np.load(os.path.join(EMBEDDINGS_PATH, 'gru_128_1_0.1_128', 'embeddings.npy'))
emb.shape
```

```python
results = compute_intrinsic_dim_global(emb)
results
```

```python

results_global = {}
for embedding_dir in tqdm(embedding_dirs):
    emb = np.load(os.path.join(EMBEDDINGS_PATH, embedding_dir, 'embeddings.npy'))
    results_global[embedding_dir] = compute_intrinsic_dim_global(emb)
results_global = pd.DataFrame(results_global).T
```
